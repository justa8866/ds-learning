# Segmentacja klientów (symulacja)

Notebook pokazuje prostą segmentację na podstawie syntetycznych danych. Dane nie opisują prawdziwych osób.

## 1. Przygotowanie danych

Tworzymy sześć przykładowych rekordów z rocznymi wydatkami i liczbą zamówień.

```python
import pandas as pd

customers = [
    {"customer_id": "C001", "annual_spend_pln": 5200, "orders_per_year": 18},
    {"customer_id": "C002", "annual_spend_pln": 4100, "orders_per_year": 9},
    {"customer_id": "C003", "annual_spend_pln": 2600, "orders_per_year": 12},
    {"customer_id": "C004", "annual_spend_pln": 900, "orders_per_year": 2},
    {"customer_id": "C005", "annual_spend_pln": 1500, "orders_per_year": 4},
    {"customer_id": "C006", "annual_spend_pln": 3800, "orders_per_year": 8},
]

customers_df = pd.DataFrame(customers)
customers_df
```

## 2. Reguły segmentacji

VIP: wydatki co najmniej 4000 PLN i minimum 12 zamówień rocznie. Regular: wydatki co najmniej 2000 PLN lub minimum 8 zamówień. Pozostali klienci trafiają do segmentu Occasional.

```python
def assign_segment(row):
    if row["annual_spend_pln"] >= 4000 and row["orders_per_year"] >= 12:
        return "VIP"
    if row["annual_spend_pln"] >= 2000 or row["orders_per_year"] >= 8:
        return "Regular"
    return "Occasional"

customers_df["segment"] = customers_df.apply(assign_segment, axis=1)
customers_df
```

## 3. Podsumowanie segmentów

Porównujemy liczebność segmentów oraz średnie wydatki i liczbę zamówień.

```python
segment_summary = (
    customers_df.groupby("segment")
    .agg(
        customer_count=("customer_id", "count"),
        average_spend_pln=("annual_spend_pln", "mean"),
        average_orders_per_year=("orders_per_year", "mean"),
    )
    .round(2)
    .sort_values("customer_count", ascending=False)
)
segment_summary
```
